```python
from pathlib import Path
import duckdb
import pandas as pd

db_path = Path("data/fpl.duckdb")
if not db_path.exists():
    db_path = Path("../data/fpl.duckdb")

con = duckdb.connect(str(db_path), read_only=True)
```

```python
recommendations = con.execute("""
    SELECT
        player_name,
        team_name,
        position_id,
        current_price_tenths / 10.0 AS price_million,
        projected_points,
        posterior_ppm,
        reliability_score,
        fixture_difficulty,
        recommendation_score
    FROM player_recommendation
    ORDER BY recommendation_score DESC
    LIMIT 50
""").df()

recommendations
```

```python
import plotly.express as px

fig = px.scatter(
    recommendations,
    x="price_million",
    y="projected_points",
    color="position_id",
    size="reliability_score",
    hover_name="player_name",
    hover_data=["team_name", "posterior_ppm", "fixture_difficulty"],
    title="Projected points vs price"
)

fig.show()
```
